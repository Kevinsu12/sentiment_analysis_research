```python
import pandas as pd
```

```python
PATH = 'data/earning_call_presentations/tr_sp500_final_2000s.parquet'
PATH_CLEAN = 'data/earning_call_presentations/clean_tr_sp500_final_2000s.parquet'
```

```python
df = pd.read_parquet(PATH)
```

```python
# df = df.sort_values(by = ['transcriptid','componentorder']).drop_duplicates(subset = ['transcript_text'])
```

```python
# df.to_parquet(PATH)
# df
```

```python
counts = df.groupby('transcriptid')['mostimp_etc'].nunique() 
faulty = counts[counts > 1].index.to_list()
```

```python
counts
```

```python
len(faulty)
```

```python
len(faulty) / len(df['transcriptid'].drop_duplicates()) 
```

```python
df = df[~df['transcriptid'].isin(faulty)]
```

```python
df.shape
```

```python
ids = df.groupby(['permno', 'mostimp_etc', 'transcriptid']).size().reset_index(name = 'size')
ids = ids.sort_values('size', ascending=False).drop_duplicates(subset=['permno', 'mostimp_etc'])['transcriptid'].to_list()
```

```python
df = df[df['transcriptid'].isin(ids)]
```

```python
df.shape
```

```python
df.to_parquet(PATH_CLEAN)
```
